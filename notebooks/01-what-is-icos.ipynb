# 01 — What is ICOS?

**Learning goal:** Identify what ICOS measures, what a station is, and where to find its data. 


## One system, three kinds of observations

The **Integrated Carbon Observation System (ICOS)** is a European network of measurement stations. Its long-term observations help people study greenhouse gases and the carbon cycle. Its three domains are:

| Domain | Plain-language question | Example observation |
| --- | --- | --- |
| **Atmosphere** | How much of a gas is present in the air? | Carbon dioxide (CO₂) concentration in air. |
| **Ecosystem** | How do land ecosystems and the air exchange gases? | CO₂ moving between a peatland and the air. |
| **Ocean** | How do the ocean and the air exchange carbon? | Carbon observations at sea. |

A **concentration** says how much gas is in a sample of air. A **flux** describes the rate of gas exchange across a surface over an area and time. They answer different questions and do not use the same units. [Source: ICOS in a nutshell](https://www.icos-cp.eu/about/icos-in-nutshell).


## What is a station?

A station is a measurement site or platform in one of the three domains. An Ecosystem station often uses a tower and nearby instruments to observe gas exchange, weather, soil and vegetation. Its measurements describe the surrounding area influencing the instruments; they are **not a map of every peatland pixel**. A station page describes the site. A linked **data-object page** describes a particular published product, its period and its fields. Having a station page alone does not show that every gas or variable is available for every year.

Browse the official [station network](https://www.icos-cp.eu/measurements/station-network) or read about [Ecosystem measurements](https://www.icos-cp.eu/measurements/ecosystem). The network directory can be searched by domain and country.


## From station to data product

**Station observations → Thematic Centre processing and quality checks → Carbon Portal product page → reader**

For an Ecosystem station, the **Ecosystem Thematic Centre (ETC)** coordinates processing and quality control. The **Carbon Portal** makes released products and their descriptions available to readers. Different releases may cover different periods and variables. [Sources: ICOS data flow](https://www.icos-cp.eu/measurements/data-collection/data-flow); [Ecosystem measurements](https://www.icos-cp.eu/measurements/ecosystem).

## Remember

- ICOS links Atmosphere, Ecosystem and Ocean observations.
- Concentration is an amount in air; flux is a rate of exchange.
