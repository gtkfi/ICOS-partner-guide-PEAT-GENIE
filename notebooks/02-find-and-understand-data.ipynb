# 02 — Find and understand ICOS data

**Learning goal:** Locate one Ecosystem product and read its description before using its numbers. Reading time: about 7 minutes. This guide does not download data.


## Browse in a browser

1. Open the [ICOS Carbon Portal guide](https://www.icos-cp.eu/data-services/data-portal/how-to-use) and follow its **Data Portal** link. The portal has a **Filters** menu and search results; the layout may change.
2. Filter to the **Ecosystem** domain and choose a station, such as **Lompolojankka (FI-Lom)**. You can also start in the [station network](https://www.icos-cp.eu/measurements/station-network), open a station page, and follow **Browse data**.
3. Open one matching data object. Read its title, station, data type, time coverage, data level, licence, and persistent identifier (**PID**) or digital object identifier (**DOI**) if supplied. Check the **release** and file version too. A station can have several products.
4. Inspect its field names and **units**. If the object offers a **Preview** control, use it to see a variable without downloading. Preview is not available for every object.

The portal's [official instructions](https://www.icos-cp.eu/data-services/data-portal/how-to-use) explain filters, landing pages and previews.

## What do ICOS data levels mean?

| Label | Plain meaning | What to remember |
| --- | --- | --- |
| **L0** | Raw observation | Not yet fully quality checked; access to raw data can be restricted or on request. |
| **L1** | Intermediate observations used during processing | Ordinary L1 working data are normally internal to ICOS. |
| **NRT (special L1)** | Near-real-time data released quickly after automated checks | Useful for recent conditions; it is not the final L2 release. |
| **L2** | Final quality-controlled observations | ICOS's main final observational product. |
| **L3** | Elaborated product built partly or wholly from ICOS data | May be contributed by groups outside ICOS; read its methods. |

[Source: ICOS data levels and quality](https://www.icos-cp.eu/measurements/data-collection/data-levels-quality). **NRT** means *near-real-time*. A number's quality level does not tell you whether a particular gas, station or year is present: open the exact object page.


## Licence and citation

Reading product descriptions in this guide requires no account. If you later download data, the portal asks you to accept the applicable data licence; an account can remember that acceptance. The [ICOS Data Licence](https://www.icos-cp.eu/data-services/data-portal/data-licence) explains attribution under CC BY 4.0.

For a report or publication, copy the **citation from the exact data-object page**, including its PID or DOI as shown, and acknowledge ICOS where appropriate. If you use several objects from one release, follow [ICOS's citation guide](https://www.icos-cp.eu/how-to-cite) for the release citation. A station URL by itself is not a substitute for citing the product used.


## Remember

- A station and a data object are different pages: use the object to verify period, variables, units and level.
- NRT is fast special L1; L2 is the final quality-controlled observation product.
- Save the object's release if listed, identifier and supplied citation before using its data.
