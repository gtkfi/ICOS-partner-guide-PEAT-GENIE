# 04 · ICOS feature glossary

**Learning goal:** Read common labels on Ecosystem data pages without treating every data product as if it had the same columns.


## First, separate ideas from columns

| Term | Plain meaning | Example unit | Product context | Interpretation or quality note | Official source |
| --- | --- | --- | --- | --- | --- |
| EC (eddy covariance) | A tower method that uses fast air and gas measurements to estimate gas exchange between a surface and the atmosphere. | Method; no unit | Ecosystem observations | The measured exchange refers to a changing area upwind of the tower, not one fixed map pixel. | [ICOS Ecosystem measurements](https://www.icos-cp.eu/measurements/ecosystem) |
| NEE (net ecosystem exchange) | The net balance of CO₂ going between an ecosystem and the air. | µmol CO₂ m⁻² s⁻¹ for the half-hourly FI-Lom product | Concept and named FI-Lom Fluxnet field below | In the CO₂ flux convention illustrated by ICOS, negative means net uptake from the air, positive means net release; check the product's convention before interpreting a different flux. NEE is not the same field as `FC`. | [ICOS flux explanation](https://www.icos-cp.eu/media/184); [FI-Lom L2 object](https://meta.icos-cp.eu/objects/-v0Irkxc4r6zIW4Iqh69EUSF) |
| GPP (gross primary production) | CO₂ taken up through photosynthesis, estimated by separating a processed NEE series into components. | µmol CO₂ m⁻² s⁻¹ for FI-Lom half-hourly fields | FI-Lom Fluxnet partitioned fields | An estimate from a processing method, not a second direct tower gas measurement. Daytime and nighttime methods give separate versions. | [FI-Lom L2 object](https://meta.icos-cp.eu/objects/-v0Irkxc4r6zIW4Iqh69EUSF); [FLUXNET variable guide](https://fluxnet.org/data/fluxnet2015-dataset/variables-quick-start-guide/) |
| RECO (ecosystem respiration) | CO₂ released by living organisms and their decomposition processes, estimated as a component of NEE. | µmol CO₂ m⁻² s⁻¹ for FI-Lom half-hourly fields | FI-Lom Fluxnet partitioned fields | Daytime (`DT`) and nighttime (`NT`) estimates can differ. | [FI-Lom L2 object](https://meta.icos-cp.eu/objects/-v0Irkxc4r6zIW4Iqh69EUSF); [FLUXNET variable guide](https://fluxnet.org/data/fluxnet2015-dataset/variables-quick-start-guide/) |
| L2 (Level 2) | ICOS's final, quality-checked observational data level. | Data level; no unit | ICOS product metadata | Still inspect the particular product's processing, gaps and flags; “L2” is not a variable name. | [ICOS data levels](https://www.icos-cp.eu/measurements/data-collection/data-levels-quality) |

Here, **flux** means gas passing through an area over time. A concentration in air is a different measurement. The example flux unit “µmol m⁻² s⁻¹” means micromoles per square metre per second.


## Then, read the actual product fields

ETC means **Ecosystem Thematic Centre**. The table draws from ETC Fluxes and Meteo examples and the **FI-Lom ETC L2 Fluxnet half-hourly** product, which uses the ONEFlux processing procedure. The methane example is a **different Level 1 product from the Swedish National Network**, tagged pre-ICOS in the portal. It is included only to show how a methane field may look; it is not an ETC L2 field. Names and units can change by product family.

| `TIMESTAMP`, `TIMESTAMP_END` | Time labels for the record and interval end. | Time label | ETC L2 Fluxnet half-hourly, FI-Lom | The landing page lists `TIMESTAMP` as UTC and `TIMESTAMP_END` as local non-daylight-saving time; check time definitions before joining other files. | [FI-Lom L2 object](https://meta.icos-cp.eu/objects/-v0Irkxc4r6zIW4Iqh69EUSF) |
| `FC` | Carbon dioxide (CO₂) flux from eddy covariance processing. | µmol m⁻² s⁻¹ | ETC L2 Fluxes example, Kumpula | A flux field; do not substitute it for processed/gap-filled `NEE_VUT_REF`. | [Kumpula L2 Fluxes object](https://meta.icos-cp.eu/objects/jQIA4EkPznVGZF2w1cIqjAoO); [ETC flux variable list](https://meta.icos-cp.eu/objects/E0KJJERYMTDbg4PfW26xazhk) |

## A short reading example

On the [FI-Lom L2 Fluxnet half-hourly data page](https://meta.icos-cp.eu/objects/-v0Irkxc4r6zIW4Iqh69EUSF), find `NEE_VUT_REF` in **Previewable variables**. Its unit is µmol m⁻² s⁻¹. Find `NEE_VUT_REF_QC` on the same page to see that a separate quality column accompanies that exchange series. The page identifies the station, time span, product level and persistent identifier. The value's sign describes net CO₂ exchange in this convention; a negative value corresponds to uptake by the ecosystem, a positive value to release. A half-hourly value does **not** by itself describe an annual carbon balance, and CO₂ exchange does not include methane.

**Remember**

1. Check the **exact data-object page** before assuming a column exists or has a given unit.
2. `FC` is a CO₂ flux field; `NEE_VUT_REF` is a processed exchange series. GPP and RECO are partitioned estimates.
3. A quality flag and a water-table sign convention need their **own product documentation**. Methane has its own fields and may not appear in a CO₂ product.
