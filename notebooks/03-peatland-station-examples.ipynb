# 03. Peatland station examples

**Learning goal:** Read an ICOS Ecosystem station page, identify why a site is relevant to peatlands, and find its data products.

A **station** is a measurement site. Its page describes the place and links to data published through the ICOS Carbon Portal. The station's *main ecosystem* is a classification label; read the site description as well. These four examples are a starting point, not a complete list of ICOS peatland sites.

**Why Lettosuo matters:** Its description says *peatland forest*, but its main ecosystem is *Evergreen Needleleaf Forests*. Searching only for “Permanent Wetlands” can miss peatland sites.


## From a station page to a data product

1. Open one linked station page and read its **Description** and **Main ecosystem**.
2. Select **Browse data** or **Data** on that page. This opens the Carbon Portal results for the station.
3. Open a product of interest and read its **product type**, **time span**, **data level**, **version or release**, and **persistent identifier (PID) or DOI** if shown.
4. Check the product documentation for its actual variables and units. If you intend to use it, also check the licence and citation instructions.

A station page confirms the site's identity and surroundings. It does **not** promise a particular CO₂ flux, CH₄ flux, water-table field, time span, or quality level. Products differ by station and release. Recheck the Carbon Portal when you need specific data. A tower's flux measurement is influenced by the area upwind of the tower at that time; it is not a map of every peatland pixel nearby.


## Remember

- Use the **site description and ecosystem label together** when looking for peatlands.
- Follow **Browse data** and inspect each product; do not assume that every station measures the same variables.
- Record the **station ID, product identifier, and date checked** when sharing a data source.
